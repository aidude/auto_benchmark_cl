# auto_benchmark_cl — Results Analysis
Load results from `results/` and compute Gain / BWT / FWT.

```python
import json
import pathlib
import pandas as pd

results_dir = pathlib.Path('../results')
records = []
for f in results_dir.rglob('*.json'):
    data = json.loads(f.read_text())
    records.append(data)

df = pd.DataFrame(records)
df.head()
```

```python
# Pivot: model × task accuracy
if not df.empty:
    pivot = df.pivot_table(index='model', columns='task', values='accuracy')
    display(pivot.style.background_gradient(cmap='RdYlGn', vmin=0, vmax=1))
```
